```python
employees = [
    {"id": 101, "name": "Alice", "department": "Data", "salary": 65000, "rating": 4.5},
    {"id": 102, "name": "Bob", "department": "AI", "salary": 72000, "rating": 4.8},
    {"id": 103, "name": "Charlie", "department": "Web", "salary": 58000, "rating": 3.9},
    {"id": 104, "name": "David", "department": "Data", "salary": 61000, "rating": 4.2},
    {"id": 105, "name": "Eva", "department": "AI", "salary": 75000, "rating": 4.9},
    {"id": 106, "name": "Frank", "department": "Cloud", "salary": 68000, "rating": 4.1},
    {"id": 107, "name": "Grace", "department": "Web", "salary": 59000, "rating": 3.8},
    {"id": 108, "name": "Helen", "department": "Data", "salary": 70000, "rating": 4.6}
]
#total employess
print("Total Employees:", len(employees))

#  Salary statistics
salaries = []

for employee in employees:
    salaries.append(employee["salary"])

print("Average Salary:", sum(salaries) / len(salaries))
print("Highest Salary:", max(salaries))
print("Lowest Salary:", min(salaries))


 #Employees with salary > 65000 and rating >= 4.5

print("Employees with salary > 65000 and rating >= 4.5:")
for employee in employees:
    if employee["salary"] > 65000 and employee["rating"] >= 4.5:
        print(employee["name"])


#Employee ID - Name dictionary
employee_dict = {}

for employee in employees:
    employee_dict[employee["id"]] = employee["name"]

print("Employee ID - Name Dictionary:")
print(employee_dict)


#  Employee names

name_list = []

for employee in employees:
    name_list.append(employee["name"])
name_list.sort()
name_list.reverse()

print("\nFirst Three Employee Names")
for name in name_list[:3]:
    print(name)


#  Tuple of Employee IDs, max and min
id_tuple = tuple(emp["id"] for emp in employees)
print("\nEmployee ID Tuple")
print(id_tuple)
print("Maximum Employee ID:", max(id_tuple))
print("Minimum Employee ID:", min(id_tuple))

# 7. Unique departments
dept_set = set()
for emp in employees:
    dept_set.add(emp["department"])

print("\nDepartments")
print(dept_set)

# 8. Search employee by ID
emp_id = int(input("\nEnter Employee ID: "))

found = False
for emp in employees:
    if emp["id"] == emp_id:
        found = True
        print("\nEmployee Found")
        print("ID:", emp["id"])
        print("Name:", emp["name"])
        print("Department:", emp["department"])
        print("Salary:", emp["salary"])
        print("Rating:", emp["rating"])
        break

if not found:
    print("\nEmployee Not Found")
```
